# Bài 3.4



```python
import statistics
x = [1, 2, 3, 4, 5]
y = [4.0, 5.5, 6.0, 7.5, 8.0]
n = len(x)
tb_x = sum(x) / n
tb_y = sum(y) / n

tu_so = 0
mau_so = 0
for xi, yi in zip(x, y):
    tu_so += (xi - tb_x) * (yi - tb_y)
    mau_so += (xi - tb_x)**2
b = tu_so / mau_so
a = tb_y - b * tb_x
print(f"Phương trình: y = {a:.2f} + {b:.2f}x")
print(f"Điểm dự đoán 6 giờ ôn: {a + b * 6:.2f}")
kq = statistics.linear_regression(x, y)
print("Kết quả:", kq)
```

# Bài 3.6

```python
from collections import defaultdict
danh_sach = [("An", 9.0), ("Binh", 7.5), ("Chi", 8.6), ("Dung", 5.0), ("Em", 6.0), ("Phuc", 3.5), ("Giang", 7.0)]
nhom_diem = defaultdict(list)
for ten, diem in danh_sach:
    if diem >= 8.5:
        diem_chu = 'A'
    elif diem >= 7:
        diem_chu = 'B'
    elif diem >= 5.5:
        diem_chu = 'C'
    elif diem >= 4:
        diem_chu = 'D'
    else:
        diem_chu = 'F'
    nhom_diem[diem_chu].append(ten)
for diem_chu in sorted(nhom_diem.keys()):
    print(f"Điểm {diem_chu}: {nhom_diem[diem_chu]}")
```

# bài 3.8

```python
cua_hang_1 = {"ao": 120, "quan": 80, "non": 30}
cua_hang_2 = {"ao": 90, "giay": 150, "non": 20}
tong_doanh_thu = dict(cua_hang_1)

for san_pham, tien in cua_hang_2.items():
    if san_pham in tong_doanh_thu:
        tong_doanh_thu[san_pham] += tien
    else:
        tong_doanh_thu[san_pham] = tien
doanh_thu_giam_dan = dict(sorted(tong_doanh_thu.items(), key=lambda x: x[1], reverse=True))
print("Tổng doanh thu giảm dần:", doanh_thu_giam_dan)
tap_hop_1 = set(cua_hang_1.keys())
tap_hop_2 = set(cua_hang_2.keys())
ban_o_ca_hai = tap_hop_1 & tap_hop_2
print("Bán ở cả hai cửa hàng:", ban_o_ca_hai)
ban_o_mot_noi = tap_hop_1 ^ tap_hop_2
print("Chỉ bán ở một cửa hàng:", ban_o_mot_noi)
```

# bài 3.10

```python
n = 1000
snt = set(range(2, n ))
p = 2
while p * p <= n:
    if p in snt:
        for k in range(p * p, n , p):
            snt.discard(k)
    p += 1
print("Số lượng số nguyên tố:", len(snt))
print("Tổng các số nguyên tố:", sum(snt))
dem = 0
for p in snt:
    if p + 2 in snt:
        dem += 1
print("Số cặp số nguyên tố sinh đôi:", dem)
```
